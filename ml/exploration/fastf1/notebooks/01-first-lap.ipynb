```python
import os
import fastf1

# cache downloads to fastf1-cache instead of default folder Temp, which Windows cleanup tools may empty.
CACHE = os.path.join(os.environ["LOCALAPPDATA"], "fastf1-cache")
os.makedirs(CACHE, exist_ok=True)
fastf1.Cache.enable_cache(CACHE)

# load session
session = fastf1.get_session(2021, 7, 'Q')
session
```

```python
session.load()
car_data = session.laps.pick_fastest().get_car_data()
```

```python
car_data = car_data.drop(columns=["nGear", "Source", "DRS", "Date"])
car_data["Brake"] = car_data["Brake"].astype('int')
car_data["Time"] = car_data["Time"].dt.total_seconds()

car_data
```
